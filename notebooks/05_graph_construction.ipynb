# 05 Graph Construction — SENTINEL
Output for 06: `artifacts/graph.gpickle` + `*_fe.csv` with graph features. Train-only graph (no valid/test leakage).

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path('.').resolve().parent))
```

```python
import pandas as pd, pickle
from src.graph_builder import build_graph, graph_features
from src import config
```

```python
train = pd.read_csv(config.ARTIFACTS_DIR / 'train_fe.csv')
valid = pd.read_csv(config.ARTIFACTS_DIR / 'valid_fe.csv')
test = pd.read_csv(config.ARTIFACTS_DIR / 'test_fe.csv')
G = build_graph(train)
print('nodes', G.number_of_nodes(), 'edges', G.number_of_edges())
```

```python
train_g = graph_features(train, G)
valid_g = graph_features(valid, G)
test_g = graph_features(test, G)
print(train_g[['in_degree','out_degree','pagerank','fan_in_7d','fan_out_7d']].describe())
print(train_g.groupby('is_suspicious')[['pagerank','fan_in_7d','fan_out_7d','two_hop_reach']].mean())
```

```python
with open(config.ARTIFACTS_DIR / 'graph.gpickle', 'wb') as f:
    pickle.dump(G, f)
train_g.to_csv(config.ARTIFACTS_DIR / 'train_fe.csv', index=False)
valid_g.to_csv(config.ARTIFACTS_DIR / 'valid_fe.csv', index=False)
test_g.to_csv(config.ARTIFACTS_DIR / 'test_fe.csv', index=False)
print('saved graph.gpickle + *_fe.csv with graph features -> used by 06_graphsage_pretraining.ipynb')
```
