Output for 07: `artifacts/graph_emb.joblib` (zeros fallback if no torch). Link-prediction on train graph only.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path('.').resolve().parent))
```

```python
import pickle, joblib, numpy as np
from src import graph_model, config
print('torch available:', graph_model.AVAILABLE, 'emb_dim:', graph_model.EMB_DIM)
```

```python
with open(config.ARTIFACTS_DIR / 'graph.gpickle', 'rb') as f:
    G = pickle.load(f)
print('nodes', G.number_of_nodes(), 'edges', G.number_of_edges())
model = graph_model.train_link_prediction(G)
print('model:', type(model).__name__ if model is not None else 'None -> dummy zeros')
```

```python
emb = graph_model.get_embeddings(G, model)
print('n_emb', len(emb))
import random
for k in random.sample(sorted(emb)[:100] if len(emb)>100 else sorted(emb), min(2, len(emb))):
    print(k, np.asarray(emb[k]).shape, np.asarray(emb[k])[:4])
```

```python
joblib.dump(emb, config.ARTIFACTS_DIR / 'graph_emb.joblib')
print('saved graph_emb.joblib -> used by 07_feature_fusion.ipynb')
```
