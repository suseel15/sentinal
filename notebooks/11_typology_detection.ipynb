9 typologies with evidence via `src/typology_engine.identify_typology`.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path('.').resolve().parent))
```

```python
import pandas as pd
from collections import Counter
from src.typology_engine import identify_typology
from src import config
```

```python
test = pd.read_csv(config.ARTIFACTS_DIR / 'test_fe.csv')
test['pred_typos'] = test.apply(lambda r: identify_typology(r.to_dict(), names_only=True), axis=1)
print(Counter([t for xs in test['pred_typos'] for t in xs]))
```

```python
# One evidence example per typology (9 total)
seen = set()
for _, r in test.iterrows():
    for d in identify_typology(r.to_dict()):
        if d['name'] not in seen:
            print(d['name'], '|', d['confidence'], '|', d['evidence'])
            seen.add(d['name'])
    if len(seen) >= 9:
        break
print('typologies found:', sorted(seen))
```
