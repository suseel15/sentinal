```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path('.').resolve().parent))
```

```python
import numpy as np
from scipy.sparse import load_npz
from src.xgboost_model import train_xgb, tune_thresholds, predict_proba
from src import config
```

```python
Xtr = load_npz(config.ARTIFACTS_DIR / 'X_train.npz').toarray()
Xva = load_npz(config.ARTIFACTS_DIR / 'X_valid.npz').toarray()
ytr = np.load(config.ARTIFACTS_DIR / 'y_train.npy')
yva = np.load(config.ARTIFACTS_DIR / 'y_valid.npy')
print(Xtr.shape, Xva.shape, ytr.mean(), yva.mean())
model = train_xgb(Xtr, ytr, Xva, yva)
print('saved xgb.json')
```

```python
probs = predict_proba(model, Xva)
thr = tune_thresholds(yva, probs)
print(thr)
```

```python
from sklearn.metrics import average_precision_score
print('valid PR-AUC', average_precision_score(yva, probs))
print('artifacts xgb.json + thresholds.json -> used by 09_model_evaluation.ipynb')
```
